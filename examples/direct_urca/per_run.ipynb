```python
from __future__ import annotations

import os
import logging
from pathlib import Path
from collections import defaultdict

import h5py
import jax
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.collections import LineCollection
from matplotlib.lines import Line2D
from matplotlib.ticker import MaxNLocator
from IPython.display import Markdown, display

# =============================================================================
# JESTER & UTILS
# =============================================================================
jax.config.update("jax_platforms", "cpu")
try:
    from jesterTOV import utils
except ImportError:
    class utils:  # type: ignore[no-redef]
        fm_inv3_to_SI = 1e45
        geometric_to_MeV_fm_inv3 = 1.0
        MeV_fm_inv3_to_geometric = 1.0
        fm_inv3_to_geometric = 1.0

# =============================================================================
# CONFIGURATION
# =============================================================================
logging.basicConfig(level=logging.INFO, format="%(message)s")
logger = logging.getLogger("eos_master")

PLOT_CONFIG = {
    "DPI": 150,
    "USE_TEX": True,
    "FIG_SIZE": (14, 6),
    "NUM_SAMPLES": 1000,
    "ALPHA": 0.15,
    "LINEWIDTH": 1.0,
    "FIG_WIDTH": 16,
    "COLS": 4,
    "BINS": 40,
}

try:
    CURRENT_DIR = Path(__file__).resolve().parent
except NameError:
    CURRENT_DIR = Path.cwd()

ROOT_JESTER = CURRENT_DIR.parents[1]
BASE_PATH = ROOT_JESTER / "jesterTOV" / "inference"

CHIEFT_CONFIG = {
    "DATA_DIR": BASE_PATH / "data" / "chiEFT" / "2402.04172",
    "LOW_FILE": "low.dat",
    "HIGH_FILE": "high.dat",
    "COLOR": "gray",
    "ALPHA": 0.7,
    "ZORDER": 0,
}

N_SAT_FM = 0.16
N_SAT_SI = N_SAT_FM * utils.fm_inv3_to_SI
GEOM_TO_MEVFM3 = utils.geometric_to_MeV_fm_inv3

def setup_matplotlib():
    plt.rcParams.update(
        {
            "axes.grid": True,
            "grid.alpha": 0.3,
            "axes.labelsize": 14,
            "xtick.labelsize": 12,
            "ytick.labelsize": 12,
            "legend.fontsize": 12,
            "figure.titlesize": 16,
            "mathtext.fontset": "cm",
            "font.family": "serif",
            "text.color": "#000000",
            "axes.labelcolor": "#000000",
            "xtick.color": "#000000",
            "ytick.color": "#000000",
        }
    )
    if PLOT_CONFIG["USE_TEX"]:
        try:
            from matplotlib.checkdep_usetex import checkdep_usetex
            if checkdep_usetex(True):
                plt.rcParams.update({"text.usetex": True})
        except Exception:
            pass

def _normalize_input(item):
    if isinstance(item, dict):
        return item["path"], item.get("filters")
    if isinstance(item, (tuple, list)) and len(item) == 2:
        return item
    return item, None

# ==============================================================================
# H5 DATA LOADING
# ==============================================================================

def load_h5_data(filepath, mode="curves", filters=None):
    if not os.path.exists(filepath):
        logger.warning(f"File not found: {filepath}")
        return None

    with h5py.File(filepath, "r") as f:
        if "posterior" not in f:
            return None
        post = f["posterior"]

        n_samples = 0
        ref_param = None

        if filters:
            for k in filters.keys():
                if "parameters" in post and k in post["parameters"]:
                    n_samples = post["parameters"][k].shape[0]
                    ref_param = k
                    break
                if "derived_eos" in post and k in post["derived_eos"]:
                    n_samples = post["derived_eos"][k].shape[0]
                    ref_param = k
                    break

        if n_samples == 0:
            if "parameters" in post:
                ref_param = list(post["parameters"].keys())[0]
                n_samples = post["parameters"][ref_param].shape[0]
            elif "derived_eos" in post:
                grp = post["derived_eos"]
                k = "masses_EOS" if "masses_EOS" in grp else "masses"
                n_samples = grp[k].shape[0]

        if n_samples == 0:
            return None

        mask = np.ones(n_samples, dtype=bool)

        if filters:
            for param, bounds in filters.items():
                dset = None
                if "parameters" in post and param in post["parameters"]:
                    dset = post["parameters"][param]
                if dset is None and "derived_eos" in post and param in post["derived_eos"]:
                    dset = post["derived_eos"][param]

                if dset is None:
                    continue
                if dset.shape[0] != n_samples:
                    continue

                data_vec = dset[:]
                mask = mask & ((data_vec >= bounds[0]) & (data_vec <= bounds[1]))

        if np.sum(mask) == 0:
            return None

        if mode == "curves":
            group = post.get("derived_eos", post)
            if "masses" not in group and "masses_EOS" not in group:
                return None
            try:
                m = group.get("masses_EOS", group.get("masses"))[:]
                r = group.get("radii_EOS", group.get("radii"))[:]
                n = group.get("n")[:] / N_SAT_SI
                p = group.get("p")[:] * GEOM_TO_MEVFM3

                durca = (
                    group.get("durca_density.nb_durca")[:]
                    if "durca_density.nb_durca" in group
                    else np.full(len(m), np.nan)
                )
                n_orig = group.get("n_orig")[:] / N_SAT_SI if "n_orig" in group else None
                e_frac = group.get("e_fraction")[:] if "e_fraction" in group else None
                mu_frac = group.get("muon_fraction")[:] if "muon_fraction" in group else None
                p_frac = group.get("proton_fraction")[:] if "proton_fraction" in group else None

                if len(m) > n_samples:
                    m, r, n, p = m[:n_samples], r[:n_samples], n[:n_samples], p[:n_samples]
                    durca = durca[:n_samples]
                    if n_orig is not None:
                        n_orig, e_frac, mu_frac, p_frac = (
                            n_orig[:n_samples],
                            e_frac[:n_samples],
                            mu_frac[:n_samples],
                            p_frac[:n_samples],
                        )
                elif len(m) < n_samples:
                    return None

                final_data = {
                    "m": [], "r": [], "n": [], "p": [], "durca": [],
                    "n_orig": [], "e_frac": [], "mu_frac": [], "p_frac": [],
                }
                indices = np.where(mask)[0]

                for i in indices:
                    if len(m[i]) > 1:
                        final_data["m"].append(m[i])
                        final_data["r"].append(r[i])
                        final_data["n"].append(n[i])
                        final_data["p"].append(p[i])
                        final_data["durca"].append(durca[i])

                        if n_orig is not None:
                            final_data["n_orig"].append(n_orig[i])
                            final_data["e_frac"].append(e_frac[i])
                            final_data["mu_frac"].append(mu_frac[i])
                            final_data["p_frac"].append(p_frac[i])

                if not final_data["m"]:
                    return None
                return final_data
            except Exception as e:
                logger.error(f"Failed to read curves: {e}")
                return None

        elif mode == "params":
            data = {}
            if "parameters" in post:
                for k in post["parameters"].keys():
                    dset = post["parameters"][k]
                    if dset.shape[0] == n_samples:
                        data[k] = dset[:][mask]

            eos_group = post.get("derived_eos", post)

            def add_derived(name, arr):
                if len(arr) >= n_samples:
                    data[name] = arr[:n_samples][mask]

            if "durca_density.nb_durca" in eos_group:
                add_derived("n_durca", eos_group["durca_density.nb_durca"][:])

            if "M_TOV" not in data:
                try:
                    masses = eos_group.get("masses_EOS", eos_group.get("masses"))[:]
                    if len(masses) >= n_samples:
                        m_tov = np.array([np.max(m_arr) if len(m_arr) > 0 else 0 for m_arr in masses])
                        add_derived("M_TOV", m_tov)
                except Exception:
                    pass

            if "R_1.4" not in data:
                try:
                    masses = eos_group.get("masses_EOS", eos_group.get("masses"))[:]
                    radii = eos_group.get("radii_EOS", eos_group.get("radii"))[:]
                    if len(masses) >= n_samples:
                        r14 = []
                        for i in range(len(masses)):
                            m_arr, r_arr = masses[i], radii[i]
                            if len(m_arr) > 1:
                                srt = np.argsort(m_arr)
                                r14.append(np.interp(1.4, m_arr[srt], r_arr[srt]))
                            else:
                                r14.append(np.nan)
                        add_derived("R_1.4", np.array(r14))
                except Exception:
                    pass
            return data
    return None

def get_segments(x_data, y_data, n_samples):
    total = len(x_data)
    if total == 0:
        return []
    sample_count = min(total, n_samples)
    indices = np.random.choice(total, sample_count, replace=False)

    segments = []
    for i in indices:
        x, y = x_data[i], y_data[i]
        mask = np.isfinite(x) & np.isfinite(y) & (x > 0)
        if np.sum(mask) > 1:
            segments.append(np.column_stack([x[mask], y[mask]]))

    return segments

def get_segments_and_colors(x_data, y_data, durca_data, base_color, n_samples):
    total = len(x_data)
    if total == 0:
        return [], []
    sample_count = min(total, n_samples)
    indices = np.random.choice(total, sample_count, replace=False)

    segments = []
    colors = []
    for idx, i in enumerate(indices):
        x, y = x_data[i], y_data[i]
        mask = np.isfinite(x) & np.isfinite(y) & (x > 0)
        if np.sum(mask) > 1:
            segments.append(np.column_stack([x[mask], y[mask]]))
            if durca_data is not None and not np.isnan(durca_data[i]):
                colors.append("red")
            else:
                colors.append(base_color)

    return segments, colors

def get_fancy_label(name):
    mapping = {
        "M_TOV": r"$M_{\rm{TOV}}$ $[M_{\odot}]$",
        "R_1.4": r"$R_{1.4}$ [km]",
        "P_3nsat": r"$P(3n_{\rm{sat}})$ [MeV fm$^{-3}$]",
        "L_sym": r"$L_{\rm{sym}}$",
        "K_sat": r"$K_{\rm{sat}}$",
        "chi2": r"$\chi^2$",
        "log_prob": r"$\ln \mathcal{L}$",
        "n_durca": r"$n_{\rm{dUrca}}$ [fm$^{-3}$]",
    }
    if name in mapping:
        return mapping[name]
    parts = name.split("_")
    base = parts[0]
    if base in ["alpha", "beta", "gamma", "delta", "rho", "sigma", "kappa", "lambda", "epsilon"]:
        base = f"\\{base}"
    if len(parts) > 1:
        joined = r"\_".join(parts[1:])
        return f"${base}_{{{joined}}}$"
    return f"${base}$"

# ==============================================================================
# PLOTTING FUNCTIONS
# ==============================================================================

def plot_spaghetti(df: pd.DataFrame, title: str):
    setup_matplotlib()
    fig, (ax_eos, ax_mr) = plt.subplots(1, 2, figsize=PLOT_CONFIG["FIG_SIZE"], dpi=PLOT_CONFIG["DPI"])

    try:
        low_path = CHIEFT_CONFIG["DATA_DIR"] / CHIEFT_CONFIG["LOW_FILE"]
        high_path = CHIEFT_CONFIG["DATA_DIR"] / CHIEFT_CONFIG["HIGH_FILE"]
        if low_path.exists() and high_path.exists():
            low = np.loadtxt(low_path)
            high = np.loadtxt(high_path)
            min_len = min(len(low), len(high))
            ax_eos.fill_between(
                low[:min_len, 0] / N_SAT_FM,
                low[:min_len, 1],
                high[:min_len, 1],
                color=CHIEFT_CONFIG["COLOR"],
                alpha=CHIEFT_CONFIG["ALPHA"],
                zorder=0,
                label="ChiEFT",
            )
    except Exception:
        pass

    seg_gray, seg_red, seg_orange = [], [], []
    seg_mr_gray, seg_mr_red, seg_mr_orange = [], [], []

    n_samples = PLOT_CONFIG["NUM_SAMPLES"]
    df_sample = df.sample(min(len(df), n_samples)) if len(df) > n_samples else df

    count_has_durca = 0
    count_no_durca = 0

    for _, row in df_sample.iterrows():
        n_arr = np.asarray(row.get("n", []))
        p_arr = np.asarray(row.get("p", []))
        r_arr = np.asarray(row.get("radii_EOS", []))
        m_arr = np.asarray(row.get("masses_EOS", []))
        pc_arr = np.asarray(row.get("logpc_EOS", []))
        
        n_split = row.get("n_trans_val", np.nan)
        logpc_split = row.get("logpc_trans", np.nan)
        n_durca = row.get("n_durca_nsat", np.nan)
        logpc_durca = row.get("logpc_durca", np.nan)
        
        has_durca = pd.notna(n_durca) and pd.notna(n_split) and n_durca < n_split
        if has_durca:
            count_has_durca += 1
        else:
            count_no_durca += 1

        # ==========================================
        # Split EOS array (n vs p)
        # ==========================================
        mask_valid = np.isfinite(n_arr) & np.isfinite(p_arr) & (n_arr > 0)
        n_v = n_arr[mask_valid]
        p_v = p_arr[mask_valid]
        
        if len(n_v) > 1 and pd.notna(n_split):
            idx_split = np.searchsorted(n_v, n_split)
            
            # 1. CSE Branch (Orange)
            if idx_split < len(n_v):
                seg_orange.append(np.column_stack([n_v[max(0, idx_split-1):], p_v[max(0, idx_split-1):]]))
            
            # 2. Before CSE
            n_pre = n_v[:idx_split+1]
            p_pre = p_v[:idx_split+1]
            
            if len(n_pre) > 1:
                if has_durca:
                    idx_durca = np.searchsorted(n_pre, n_durca)
                    if idx_durca > 0:
                        seg_gray.append(np.column_stack([n_pre[:idx_durca+1], p_pre[:idx_durca+1]]))
                    if idx_durca < len(n_pre):
                        seg_red.append(np.column_stack([n_pre[max(0, idx_durca-1):], p_pre[max(0, idx_durca-1):]]))
                else:
                    seg_gray.append(np.column_stack([n_pre, p_pre]))
        elif len(n_v) > 1:
            seg_gray.append(np.column_stack([n_v, p_v]))

        # ==========================================
        # Split MR array (R vs M)
        # ==========================================
        mask_mr = np.isfinite(r_arr) & np.isfinite(m_arr) & np.isfinite(pc_arr)
        r_v = r_arr[mask_mr]
        m_v = m_arr[mask_mr]
        pc_v = pc_arr[mask_mr]
        
        if len(pc_v) > 1 and pd.notna(logpc_split):
            idx_split_mr = np.searchsorted(pc_v, logpc_split)
            
            # 1. CSE Branch MR (Orange)
            if idx_split_mr < len(pc_v):
                seg_mr_orange.append(np.column_stack([r_v[max(0, idx_split_mr-1):], m_v[max(0, idx_split_mr-1):]]))
                
            # 2. Before CSE MR
            r_pre = r_v[:idx_split_mr+1]
            m_pre = m_v[:idx_split_mr+1]
            pc_pre = pc_v[:idx_split_mr+1]
            
            if len(r_pre) > 1:
                if has_durca and pd.notna(logpc_durca):
                    idx_durca_mr = np.searchsorted(pc_pre, logpc_durca)
                    if idx_durca_mr > 0:
                        seg_mr_gray.append(np.column_stack([r_pre[:idx_durca_mr+1], m_pre[:idx_durca_mr+1]]))
                    if idx_durca_mr < len(pc_pre):
                        seg_mr_red.append(np.column_stack([r_pre[max(0, idx_durca_mr-1):], m_pre[max(0, idx_durca_mr-1):]]))
                else:
                    seg_mr_gray.append(np.column_stack([r_pre, m_pre]))
        elif len(pc_v) > 1:
            seg_mr_gray.append(np.column_stack([r_v, m_v]))

    lw = PLOT_CONFIG["LINEWIDTH"]
    alpha_spag = 0.5
    alpha_spag_orange = 1.0
    
    for segs, c, a, z in [(seg_gray, "gray", alpha_spag, 1), 
                          (seg_red, "red", alpha_spag, 2), 
                          (seg_orange, "orange", alpha_spag_orange, 3)]:
        if segs: ax_eos.add_collection(LineCollection(segs, colors=c, alpha=a, linewidths=lw, zorder=z))
        
    for segs, c, a, z in [(seg_mr_gray, "gray", alpha_spag, 1), 
                          (seg_mr_red, "red", alpha_spag, 2), 
                          (seg_mr_orange, "orange", alpha_spag_orange, 3)]:
        if segs: ax_mr.add_collection(LineCollection(segs, colors=c, alpha=a, linewidths=lw, zorder=z))

    ax_eos.set_xscale("log")
    ax_eos.set_yscale("log")
    ax_eos.set_xlim(1e-1, 3e1)
    ax_eos.set_ylim(1e-2, 1e4)
    ax_eos.set_xlabel(r"$n/n_{\rm{sat}}$")
    ax_eos.set_ylabel(r"$P$ [MeV fm$^{-3}$]")

    ax_mr.set_xlim(9, 16)
    ax_mr.set_ylim(0.0, 3.0)
    ax_mr.set_xlabel(r"$R$ [km]")
    ax_mr.set_ylabel(r"$M$ $[M_{\odot}]$")

    custom_lines = [
        Line2D([0], [0], color="gray", lw=2),
        Line2D([0], [0], color="red", lw=2),
        Line2D([0], [0], color="orange", lw=2)
    ]
    legend_labels = [
        r"No dUrca / $n < n_{\rm{dUrca}}$",
        r"dUrca Active ($n_{\rm{dUrca}} \leq n < n_{\rm{break}}$)",
        r"CSE Branch ($n \geq n_{\rm{break}}$)"
    ]
    fig.legend(custom_lines, legend_labels, loc='upper center', bbox_to_anchor=(0.5, 1.05), ncol=3, frameon=False)

    fig.suptitle(title, y=1.12)
    plt.tight_layout()
    plt.show()

    stats_md = f"""
**🍝 Spaghetti Plot Sample Statistics:**
* Total Samples Plotted: {len(df_sample)}
* 🔴 Models with dUrca active before CSE ($n_{{\\rm{{dUrca}}}} < n_{{\\rm{{break}}}}$): {count_has_durca}
* ⚪ Models with no dUrca before CSE: {count_no_durca}
    """
    display(Markdown(stats_md))

    # --- Professional statistics from the full dataset ---
    display_statistics_table(
        compute_durca_statistics(df),
        title=f"{title}: dUrca Statistics",
    )

def compare_eos_results(*args, labels=None):
    setup_matplotlib()
    datasets, valid_paths = [], []

    for item in args:
        fpath, f_dict = _normalize_input(item)
        d = load_h5_data(fpath, mode="params", filters=f_dict)
        if d:
            datasets.append(d)
            valid_paths.append(fpath)

    if not datasets:
        return

    if not labels:
        labels = [os.path.basename(p) for p in valid_paths]
    if len(labels) < len(datasets):
        labels += [os.path.basename(p) for p in valid_paths[len(labels):]]

    colors = (
        plt.cm.tab10(np.linspace(0, 1, 10))
        if len(datasets) <= 10
        else plt.cm.turbo(np.linspace(0, 1, len(datasets)))
    )

    all_keys = sorted(list(set().union(*(d.keys() for d in datasets))))
    valid_keys = [
        k for k in all_keys if any(k in d and d[k].ndim == 1 and np.std(d[k][np.isfinite(d[k])]) > 1e-12 for d in datasets)
    ]

    if not valid_keys:
        return

    n_plots = len(valid_keys)
    cols = PLOT_CONFIG["COLS"]
    rows = (n_plots + cols - 1) // cols

    fig, axes = plt.subplots(rows, cols, figsize=(PLOT_CONFIG["FIG_WIDTH"], rows * 3.5), dpi=PLOT_CONFIG["DPI"])
    axes = axes.flatten() if n_plots > 1 else [axes]

    for i, param in enumerate(valid_keys):
        ax = axes[i]
        for j, data in enumerate(datasets):
            if param in data:
                d = data[param][np.isfinite(data[param])]
                if len(d) == 0:
                    continue

                lbl = labels[j] if i == 0 else ""
                ax.hist(d, bins=PLOT_CONFIG["BINS"], density=True, color=colors[j], alpha=0.3, label=lbl)
                ax.hist(d, bins=PLOT_CONFIG["BINS"], density=True, color=colors[j], histtype="step", linewidth=1.5)
                ax.axvline(np.median(d), color=colors[j], linestyle="dashed", alpha=0.8)

        ax.set_title(get_fancy_label(param))
        ax.set_yticks([])
        ax.xaxis.set_major_locator(MaxNLocator(nbins=5))

    for j in range(i + 1, len(axes)):
        axes[j].axis("off")
    fig.legend(loc="upper center", ncol=min(len(datasets), 5), frameon=False, bbox_to_anchor=(0.5, 1.01))
    plt.tight_layout()
    plt.show()

def plot_particle_fractions(*args, labels=None):
    setup_matplotlib()
    datasets, valid_paths = [], []

    for item in args:
        fpath, f_dict = _normalize_input(item)
        d = load_h5_data(fpath, mode="curves", filters=f_dict)
        if d and d.get("e_frac") and len(d["e_frac"]) > 0:
            datasets.append(d)
            valid_paths.append(fpath)

    if not datasets:
        return

    if not labels:
        labels = [os.path.basename(p) for p in valid_paths]
    if len(labels) < len(datasets):
        labels += [os.path.basename(p) for p in valid_paths[len(labels):]]

    colors = (
        plt.cm.tab10(np.linspace(0, 1, 10))
        if len(datasets) <= 10
        else plt.cm.turbo(np.linspace(0, 1, len(datasets)))
    )

    fig, axes = plt.subplots(1, 3, figsize=(18, 5), dpi=PLOT_CONFIG["DPI"])
    fractions = [("e_frac", r"$Y_e$"), ("mu_frac", r"$Y_\mu$"), ("p_frac", r"$Y_p$")]

    for i, data in enumerate(datasets):
        c, lbl = colors[i], labels[i]
        
        axes[0].plot([], [], color=c, label=lbl, lw=2)

        for ax, (frac_key, ylabel) in zip(axes, fractions):
            segs, seg_colors = get_segments_and_colors(
                data["n_orig"], data[frac_key], data.get("durca"), c, PLOT_CONFIG["NUM_SAMPLES"]
            )
            if segs:
                ax.add_collection(
                    LineCollection(
                        segs,
                        colors=seg_colors,
                        alpha=PLOT_CONFIG["ALPHA"],
                        linewidths=PLOT_CONFIG["LINEWIDTH"],
                    )
                )

    has_durca = any(data.get("durca") is not None and not np.all(np.isnan(data["durca"])) for data in datasets)
    if has_durca:
        axes[0].plot([], [], color="red", label="Active dUrca", lw=2)

    for ax, (_, ylabel) in zip(axes, fractions):
        ax.set_xlabel(r"$n/n_{\rm{sat}}$")
        ax.set_ylabel(ylabel)

    axes[0].legend(loc="upper left")
    plt.tight_layout()
    plt.show()

# ==============================================================================
# PANDAS DATAFRAME TOOLS
# ==============================================================================

def load_eos_to_pandas(filepath: str | Path) -> pd.DataFrame:
    data = {}
    with h5py.File(filepath, "r") as f:
        if "posterior" not in f:
            raise KeyError(f"Group 'posterior' not found in {filepath}")

        post = f["posterior"]

        if "parameters" in post:
            for key, val in post["parameters"].items():
                data[key] = val[:]

        eos_group = post.get("derived_eos", post)
        for key, val in eos_group.items():
            arr = val[:]
            if arr.ndim == 1:
                data[key] = arr
            else:
                data[key] = list(arr)

    return pd.DataFrame(data)

def plot_curve_collection(df, x_col, y_col, param_col=None, threshold_col=None, ax=None, colors='b', color_above='orange', alpha_below=0.2, alpha_above=1.0, linewidths=0.5, **kwargs):
    if ax is None:
        fig, ax = plt.subplots(figsize=(8, 6))

    segments_below = []
    segments_above = []
    segments_all = []

    for _, row in df.iterrows():
        x = np.asarray(row[x_col])
        y = np.asarray(row[y_col])
        mask_valid = np.isfinite(x) & np.isfinite(y)

        if param_col and threshold_col and pd.notna(row[threshold_col]):
            param = np.asarray(row[param_col])
            mask_valid &= np.isfinite(param)
            if np.sum(mask_valid) < 2:
                continue

            x_v = x[mask_valid]
            y_v = y[mask_valid]
            p_v = param[mask_valid]
            t_val = row[threshold_col]

            idx_cross = np.searchsorted(p_v, t_val)
            if 0 < idx_cross < len(p_v):
                mask_below = np.arange(len(p_v)) <= idx_cross
                mask_above = np.arange(len(p_v)) >= max(0, idx_cross - 1)
            else:
                mask_below = p_v <= t_val
                mask_above = p_v >= t_val

            if np.sum(mask_below) > 1:
                segments_below.append(np.column_stack([x_v[mask_below], y_v[mask_below]]))
            if np.sum(mask_above) > 1:
                segments_above.append(np.column_stack([x_v[mask_above], y_v[mask_above]]))
        else:
            if np.sum(mask_valid) > 1:
                segments_all.append(np.column_stack([x[mask_valid], y[mask_valid]]))

    if segments_all:
        lc = LineCollection(segments_all, colors=colors, alpha=alpha_below, linewidths=linewidths, **kwargs)
        ax.add_collection(lc)
    if segments_below:
        lc_below = LineCollection(segments_below, colors=colors, alpha=alpha_below, linewidths=linewidths, **kwargs)
        ax.add_collection(lc_below)
    if segments_above:
        lc_above = LineCollection(segments_above, colors=color_above, alpha=alpha_above, linewidths=linewidths, **kwargs)
        ax.add_collection(lc_above)

    ax.autoscale_view()
    return ax

def prepare_eos_dataframe(filepath: str | Path, display_status: bool = True) -> pd.DataFrame:
    if display_status:
        display(Markdown("### Loading and Processing EOS Data"))

    df = load_eos_to_pandas(filepath)

    # ==========================================
    # FILTER ACAUSAL EOS AT START
    # ==========================================
    cs2_col = next((col for col in ["cs2_EOS", "cs2"] if col in df.columns), None)
    if cs2_col:
        def is_causal(x):
            if not isinstance(x, (list, np.ndarray)) or len(x) == 0:
                return True
            arr = np.asarray(x)
            valid = arr[np.isfinite(arr)]
            if len(valid) == 0:
                return True
            return bool(np.max(valid) <= 1.0)
        
        causal_mask = df[cs2_col].apply(is_causal)
        n_acausal = (~causal_mask).sum()
        df = df[causal_mask].copy()
        df.reset_index(drop=True, inplace=True)
        if display_status and n_acausal > 0:
            display(Markdown(f"⚠️ **Causality Check**: Filtered out {n_acausal} acausal EOS models ($c_s^2 > 1$)."))

    df["p"] = df["p"] / utils.MeV_fm_inv3_to_geometric
    df["n"] = df["n"] / utils.fm_inv3_to_geometric / 0.16
    df["n_orig"] = df["n_orig"] / 0.16

    df["M_TOV"] = df["masses_EOS"].apply(max)
    idx_max = df["masses_EOS"].apply(np.argmax)

    df["R_TOV"] = [r_list[i] for r_list, i in zip(df["radii_EOS"], idx_max)]
    df["logpc_TOV"] = [pc_list[i] for pc_list, i in zip(df["logpc_EOS"], idx_max)]

    target_p_tov = np.power(10, df["logpc_TOV"]) / utils.MeV_fm_inv3_to_geometric
    df["n_TOV"] = [
        np.interp(p_target, p_list, n_list)
        for p_target, p_list, n_list in zip(target_p_tov, df["p"], df["n"])
    ]

    # Detect invalid n_TOV: NaN or at EOS table boundary (central density > EOS grid)
    n_max_per_eos = [np.max(nn[np.isfinite(nn)]) for nn in df["n"]]
    df["n_TOV_valid"] = [
        pd.notna(ntov) and ntov < n_max_eos - 0.01
        for ntov, n_max_eos in zip(df["n_TOV"], n_max_per_eos)
    ]

    # Find transition density using exact name "nbreak" first, then fallbacks
    trans_col = next((col for col in df.columns if col in ["nbreak", "n_break", "n_trans", "ntrans", "n_match"]), None)
    if trans_col:
        # The density in HDF5 is in fm^-3, so we must divide by 0.16 to get n/n_sat!
        df["n_trans_val"] = df[trans_col] / 0.16
    else:
        df["n_trans_val"] = df["n_TOV"]
        
    logpc_t = []
    for _, row in df.iterrows():
        n_list = np.array(row["n"])
        p_list = np.array(row["p"])
        nt = row["n_trans_val"]
        if pd.notna(nt) and len(n_list) > 0:
            pt = np.interp(nt, n_list, p_list)
            logpc_t.append(np.log10(pt * utils.MeV_fm_inv3_to_geometric))
        else:
            logpc_t.append(np.nan)
    df["logpc_trans"] = logpc_t

    canonical_mass = 1.4
    idx_14 = df["masses_EOS"].apply(
        lambda x: np.argmin(np.abs(np.array(x) - canonical_mass)) if np.max(x) >= canonical_mass else np.nan
    )

    df["R_14"] = [
        r_list[int(i)] if pd.notna(i) else np.nan
        for r_list, i in zip(df["radii_EOS"], idx_14)
    ]
    df["logpc_14"] = [
        pc_list[int(i)] if pd.notna(i) else np.nan
        for pc_list, i in zip(df["logpc_EOS"], idx_14)
    ]

    target_p_14 = np.power(10, df["logpc_14"]) / utils.MeV_fm_inv3_to_geometric
    df["n_14"] = [
        np.interp(p_target, p_list, n_list)
        for p_target, p_list, n_list in zip(target_p_14, df["p"], df["n"])
    ]

    df = df[(df["logpc_TOV"]) > -11.5]

    e_frac = np.array(df["e_fraction"].tolist())
    mu_frac = np.array(df["muon_fraction"].tolist())

    x_e = e_frac / (e_frac + mu_frac)
    du_frac_matrix = 1 / (1 + (1 + np.cbrt(x_e)) ** 3)
    df["DU_fraction"] = du_frac_matrix.tolist()

    ndurca_col = "durca_density.nb_durca" if "durca_density.nb_durca" in df.columns else "n_durca" if "n_durca" in df.columns else None

    yp_at_n14, yp_at_ntov = [], []
    xdu_at_n14, xdu_at_ntov = [], []
    cs2_at_n14, cs2_at_ntov, cs2_at_ndurca = [], [], []
    n_durca_vals = []
    logpc_durca_vals = []

    for _, row in df.iterrows():
        n_14_val = float(row["n_14"])
        n_break_val = float(row["n_TOV"])
        n_list = np.array(row["n"])
        p_list = np.array(row["p"])

        if all(col in df.columns for col in ["n_TOV", "n_orig", "proton_fraction", "DU_fraction"]):
            if n_14_val < n_break_val:
                yp_at_n14.append(np.interp(n_14_val, row["n_orig"], row["proton_fraction"], left=np.nan, right=np.nan))
                xdu_at_n14.append(np.interp(n_14_val, row["n_orig"], row["DU_fraction"], left=np.nan, right=np.nan))
            else:
                yp_at_n14.append(np.nan)
                xdu_at_n14.append(np.nan)

            yp_at_ntov.append(np.interp(n_break_val, row["n_orig"], row["proton_fraction"], left=np.nan, right=np.nan))
            xdu_at_ntov.append(np.interp(n_break_val, row["n_orig"], row["DU_fraction"], left=np.nan, right=np.nan))

        if cs2_col and isinstance(row[cs2_col], (list, np.ndarray)):
            cs2_list = np.array(row[cs2_col])
            cs2_at_n14.append(np.interp(n_14_val, n_list, cs2_list, left=np.nan, right=np.nan) if pd.notna(n_14_val) else np.nan)
            cs2_at_ntov.append(np.interp(n_break_val, n_list, cs2_list, left=np.nan, right=np.nan) if pd.notna(n_break_val) else np.nan)

            if ndurca_col in df.columns:
                nd_val = float(row[ndurca_col]) / 0.16 
                n_durca_vals.append(nd_val)
                cs2_at_ndurca.append(np.interp(nd_val, n_list, cs2_list, left=np.nan, right=np.nan))
                pdurca = np.interp(nd_val, n_list, p_list)
                logpc_durca_vals.append(np.log10(pdurca * utils.MeV_fm_inv3_to_geometric))
            else:
                n_durca_vals.append(np.nan)
                cs2_at_ndurca.append(np.nan)
                logpc_durca_vals.append(np.nan)
        else:
            cs2_at_n14.append(np.nan)
            cs2_at_ntov.append(np.nan)
            cs2_at_ndurca.append(np.nan)
            n_durca_vals.append(np.nan)
            logpc_durca_vals.append(np.nan)

    if "proton_fraction" in df.columns:
        df["Yp_n14"] = yp_at_n14
        df["Yp_nTOV"] = yp_at_ntov
        df["XDU_n14"] = xdu_at_n14
        df["XDU_nTOV"] = xdu_at_ntov

    if cs2_col:
        df["cs2_n14"] = cs2_at_n14
        df["cs2_nTOV"] = cs2_at_ntov
        df["cs2_ndurca"] = cs2_at_ndurca
        df["n_durca_nsat"] = n_durca_vals
        df["logpc_durca"] = logpc_durca_vals
        df["cs2_array"] = df[cs2_col]

    return df

def prepare_eos_datasets(file_specs, display_status: bool = False):
    datasets = {}
    for item in file_specs:
        if isinstance(item, dict):
            path = item["path"]
            label = item.get("label", Path(path).stem)
        else:
            path = item
            label = Path(path).stem
        datasets[label] = prepare_eos_dataframe(path, display_status=display_status)
    return datasets

# =============================================================================
# dURCA STATISTICS & PROFESSIONAL TABLE
# =============================================================================

# =============================================================================
# dURCA STATISTICS & PROFESSIONAL TABLE
# =============================================================================

def compute_durca_statistics(df: pd.DataFrame) -> dict[str, int | float]:
    """Compute comprehensive dUrca statistics from a processed EOS DataFrame,
    excluding samples with invalid n_TOV.

    Parameters
    ----------
    df : pd.DataFrame
        Processed EOS dataframe from :func:`prepare_eos_dataframe`.

    Returns
    -------
    dict[str, int | float]
        Dictionary with keys: ``total_samples``, ``total_valid``,
        ``with_durca``, ``without_durca``, ``durca_below_nbreak``,
        ``durca_between_n14_and_nbreak``, ``durca_above_nbreak``,
        and corresponding new conditional keys.
    """
    total = len(df)
    if total == 0:
        return {
            "total_raw": 0,
            "total_samples": 0,
            "with_durca": 0,
            "without_durca": 0,
            "durca_below_nbreak": 0,
            "durca_between_n14_and_nbreak": 0,
            "durca_above_nbreak": 0,
            "nbreak_gt_tov": 0,
            "durca_below_nbreak_given_lt_tov": 0,
            "durca_between_n14_and_nbreak_given_lt_tov": 0,
            "durca_above_nbreak_given_lt_tov": 0,
            "nbreak_gt_tov_and_durca_gt_n14_given_lt_tov": 0,
            "durca_gt_tov": 0,
            "frac_durca": 0.0,
            "frac_durca_below_nbreak": 0.0,
            "frac_durca_between_n14_and_nbreak": 0.0,
        }

    has_durca_mask = df["n_durca_nsat"].notna()
    no_durca_mask = df["n_durca_nsat"].isna()

    n_durca = df["n_durca_nsat"]
    n_break = df["n_trans_val"]
    n_14 = df["n_14"].fillna(0.0)
    n_tov = df["n_TOV"]

    # Keep existing features untouched
    durca_below_nbreak_mask = has_durca_mask & (n_durca < n_break)
    durca_between_n14_and_nbreak_mask = durca_below_nbreak_mask & (n_durca > n_14)
    durca_above_nbreak_mask = has_durca_mask & (n_durca >= n_break)
    nbreak_gt_tov_mask = n_break > n_tov

    # New conditional sub-populations
    durca_lt_tov_mask = has_durca_mask & (n_durca < n_tov)

    # 1. Direct Urca below nbreak given ndurca < ntov
    durca_below_nbreak_given_lt_tov_mask = durca_lt_tov_mask & (n_durca < n_break)

    # 2. Direct Urca below nbreak and above n_1.4 given ndurca < ntov
    durca_between_n14_nbreak_given_lt_tov_mask = durca_below_nbreak_given_lt_tov_mask & (n_durca > n_14)

    # 3. Direct Urca above nbreak (CSE branch only) given ndurca < ntov
    durca_above_nbreak_given_lt_tov_mask = durca_lt_tov_mask & (n_durca >= n_break)

    # 4. cases where n_break > ntov and ndurca > n_1.4 given ndurca < ntov
    nbreak_gt_tov_and_durca_gt_n14_given_lt_tov_mask = durca_lt_tov_mask & (n_break > n_tov) & (n_durca > n_14)

    # 5. cases where ndurca > ntov
    durca_gt_tov_mask = has_durca_mask & (n_durca > n_tov)

    stats: dict[str, int | float] = {
        "total_samples": total,
        "with_durca": int(has_durca_mask.sum()),
        "without_durca": int(no_durca_mask.sum()),
        "durca_below_nbreak": int(durca_below_nbreak_mask.sum()),
        "durca_between_n14_and_nbreak": int(durca_between_n14_and_nbreak_mask.sum()),
        "durca_above_nbreak": int(durca_above_nbreak_mask.sum()),
        "nbreak_gt_tov": int(nbreak_gt_tov_mask.sum()),
        # Additional exact cases requested
        "durca_below_nbreak_given_lt_tov": int(durca_below_nbreak_given_lt_tov_mask.sum()),
        "durca_between_n14_and_nbreak_given_lt_tov": int(durca_between_n14_nbreak_given_lt_tov_mask.sum()),
        "durca_above_nbreak_given_lt_tov": int(durca_above_nbreak_given_lt_tov_mask.sum()),
        "nbreak_gt_tov_and_durca_gt_n14_given_lt_tov": int(nbreak_gt_tov_and_durca_gt_n14_given_lt_tov_mask.sum()),
        "durca_gt_tov": int(durca_gt_tov_mask.sum()),
    }
    stats["frac_durca"] = stats["with_durca"] / total * 100
    stats["frac_durca_below_nbreak"] = stats["durca_below_nbreak"] / total * 100
    stats["frac_durca_between_n14_and_nbreak"] = (
        stats["durca_between_n14_and_nbreak"] / total * 100
    )
    return stats


def display_statistics_table(
    stats: dict[str, int | float],
    title: str = "dUrca Statistics",
    total_raw: int | None = None,
) -> None:
    """Display a professional statistics summary as a styled HTML table.

    Parameters
    ----------
    stats : dict[str, int | float]
        Statistics dictionary from :func:`compute_durca_statistics`.
    title : str
        Optional table title rendered as a Markdown heading.
    total_raw : int | None
        Original sample count before n_TOV filtering, for annotation.
    """
    if stats["total_samples"] == 0:
        display(Markdown(f"### {title}\n*No data available.*"))
        return

    total = stats["total_samples"]

    def pct(count: int) -> str:
        return f"{count / total * 100:.1f}"

    table_data = {
        "Metric": [
            "Total samples (valid $n_{\\rm{TOV}}$)",
            "With direct Urca (anywhere in star)",
            "Without direct Urca",
            "Direct Urca below $n_{\\rm{break}}$",
            (
                "Direct Urca below $n_{\\rm{break}}$ "
                "and above $n_{1.4}$"
            ),
            "Direct Urca above $n_{\\rm{break}}$ (CSE branch only)",
            "$n_{\\rm{break}} > n_{\\rm{TOV}}$",
            "Direct Urca below $n_{\\rm{break}}$ given $n_{\\rm{dUrca}} < n_{\\rm{TOV}}$",
            "Direct Urca below $n_{\\rm{break}}$ and above $n_{1.4}$ given $n_{\\rm{dUrca}} < n_{\\rm{TOV}}$",
            "Direct Urca above $n_{\\rm{break}}$ (CSE branch only) given $n_{\\rm{dUrca}} < n_{\\rm{TOV}}$",
            "Cases where $n_{\\rm{break}} > n_{\\rm{TOV}}$ and $n_{\\rm{dUrca}} > n_{1.4}$ given $n_{\\rm{dUrca}} < n_{\\rm{TOV}}$",
            "Cases where $n_{\\rm{dUrca}} > n_{\\rm{TOV}}$",
        ],
        "Count": [
            total,
            stats["with_durca"],
            stats["without_durca"],
            stats["durca_below_nbreak"],
            stats["durca_between_n14_and_nbreak"],
            stats["durca_above_nbreak"],
            stats["nbreak_gt_tov"],
            stats["durca_below_nbreak_given_lt_tov"],
            stats["durca_between_n14_and_nbreak_given_lt_tov"],
            stats["durca_above_nbreak_given_lt_tov"],
            stats["nbreak_gt_tov_and_durca_gt_n14_given_lt_tov"],
            stats["durca_gt_tov"],
        ],
        "Fraction [%]": [
            "100.0",
            pct(stats["with_durca"]),
            pct(stats["without_durca"]),
            pct(stats["durca_below_nbreak"]),
            pct(stats["durca_between_n14_and_nbreak"]),
            pct(stats["durca_above_nbreak"]),
            pct(stats["nbreak_gt_tov"]),
            pct(stats["durca_below_nbreak_given_lt_tov"]),
            pct(stats["durca_between_n14_and_nbreak_given_lt_tov"]),
            pct(stats["durca_above_nbreak_given_lt_tov"]),
            pct(stats["nbreak_gt_tov_and_durca_gt_n14_given_lt_tov"]),
            pct(stats["durca_gt_tov"]),
        ],
    }

    df_table = pd.DataFrame(table_data)

    display(Markdown(f"### {title}"))
    display(
        df_table.style
        .set_properties(**{"text-align": "left", "padding": "4px 8px"})
        .set_table_styles([
            {
                "selector": "th",
                "props": [
                    ("text-align", "left"),
                    ("padding", "6px 8px"),
                    ("font-weight", "bold"),
                    ("background-color", "#2c3e50"),
                    ("color", "white"),
                ],
            },
            {
                "selector": "td",
                "props": [("padding", "4px 8px")],
            },
            {
                "selector": "tr:nth-child(even)",
                "props": [("background-color", "#000000")], # KEEP THIS #000000, NOTHING WRONG, DON'T TOUCH
            },
            {
                "selector": "",
                "props": [
                    ("border-collapse", "collapse"),
                    ("border", "1px solid #ddd"),
                ],
            },
        ])
        .hide(axis="index")
    )

    # Additional annotation
    tov_label = r"$n_{\rm{TOV}}$"
    if total_raw is not None and total_raw != total:
        n_excluded = total_raw - total
        display(Markdown(
            "*Definitions:*  \n"
            r"$n_{\rm{dUrca}}$ - threshold density for direct Urca;  \n"
            r"$n_{\rm{break}}$ - CSE transition density;  \n"
            r"$n_{1.4}$ - central density of a $1.4\,M_\odot$ star;  \n"
            rf"{tov_label} - central density at the TOV mass.  \n"
            f"*Note:* {n_excluded} of {total_raw} samples had "
            rf"invalid {tov_label} (beyond EOS grid) and were excluded. "
            "All plots and statistics above are computed only on the "
            f"{total} samples with valid {tov_label}."
        ))
    else:
        display(Markdown(
            "*Definitions:*  \n"
            r"$n_{\rm{dUrca}}$ - threshold density for direct Urca;  \n"
            r"$n_{\rm{break}}$ - CSE transition density;  \n"
            r"$n_{1.4}$ - central density of a $1.4\,M_\odot$ star;  \n"
            rf"{tov_label} - central density at the TOV mass."
        ))
# ==============================================================================
# INTERNAL PLOT ROUTINES
# ==============================================================================

def _make_parameter_histograms(df: pd.DataFrame, label: str):
    """Plot parameter histograms from the filtered DataFrame (avoids HDF5 raw read)."""
    setup_matplotlib()

    # Columns to plot (scalar parameters of interest)
    param_configs = [
        ("M_TOV", r"$M_{\rm{TOV}}$ $[M_\odot]$"),
        ("R_14", r"$R_{1.4}$ [km]"),
        ("R_TOV", r"$R_{\rm{TOV}}$ [km]"),
        ("n_14", r"$n_{1.4}$ [$n_{\rm{sat}}$]"),
        ("n_TOV", r"$n_{\rm{TOV}}$ [$n_{\rm{sat}}$]"),
        ("n_durca_nsat", r"$n_{\rm{dUrca}}$ [$n_{\rm{sat}}$]"),
    ]

    # Also check for EOS parameters (nuclear empirical parameters)
    candidate_cols = [
        c for c in df.columns
        if c.endswith("_prior")
        or c.startswith("E_")
        or c.startswith("K_")
        or c.startswith("Q_")
        or c.startswith("Z_")
        or c.startswith("L_")
    ]
    for c in candidate_cols:
        if df[c].dtype in (np.float64, np.int64, "float64", "int64") and df[c].nunique() > 10:
            param_configs.append((c, get_fancy_label(c)))

    valid_configs = [
        (col, label_text)
        for col, label_text in param_configs
        if col in df.columns
    ]

    if not valid_configs:
        return

    n_plots = len(valid_configs)
    cols = PLOT_CONFIG["COLS"]
    rows = (n_plots + cols - 1) // cols

    fig, axes = plt.subplots(
        rows, cols,
        figsize=(PLOT_CONFIG["FIG_WIDTH"], rows * 3.5),
        dpi=PLOT_CONFIG["DPI"],
    )
    axes = axes.flatten() if n_plots > 1 else [axes]

    for i, (col, fancy_label) in enumerate(valid_configs):
        ax = axes[i]
        data = df[col].dropna().values.astype(float)
        data = data[np.isfinite(data)]

        if len(data) == 0:
            ax.set_title(fancy_label)
            continue

        # Remove physically meaningless values
        if col == "M_TOV":
            data = data[data > 0.01]

        ax.hist(data, bins=PLOT_CONFIG["BINS"], density=True,
                color="steelblue", alpha=0.3)
        ax.hist(data, bins=PLOT_CONFIG["BINS"], density=True,
                color="steelblue", histtype="step", linewidth=1.5)
        ax.axvline(np.median(data), color="steelblue",
                   linestyle="dashed", alpha=0.8)
        ax.set_title(fancy_label)
        ax.set_yticks([])
        ax.xaxis.set_major_locator(MaxNLocator(nbins=5))

    for j in range(i + 1, len(axes)):
        axes[j].axis("off")

    fig.suptitle(label, fontsize=16, y=0.98)
    plt.tight_layout()
    plt.show()


def _make_macroscopic_plot(df: pd.DataFrame, title: str):
    fig, axes = plt.subplots(1, 3, figsize=(20, 5))

    plot_curve_collection(
        df, "radii_EOS", "masses_EOS",
        ax=axes[0], colors='blue', color_above='orange', alpha_below=0.1, alpha_above=1.0, linewidths=0.5, zorder=3,
    )
    axes[0].set_xlim(9, 16)
    axes[0].set_ylim(0.0, 3.0)
    axes[0].set_xlabel("R [km]")
    axes[0].set_ylabel("M [M_sun]")

    plot_curve_collection(
        df, "logpc_EOS", "masses_EOS",
        ax=axes[1], colors='blue', color_above='orange', alpha_below=0.1, alpha_above=1.0, linewidths=0.5, zorder=3,
    )
    axes[1].set_xlim(-12.0, -8.5)
    axes[1].set_ylim(0.0, 3.0)
    axes[1].set_xlabel("log Pc")
    axes[1].set_ylabel("M [M_sun]")

    plot_curve_collection(
        df, "n", "p",
        ax=axes[2], colors='blue', color_above='orange', alpha_below=0.1, alpha_above=1.0, linewidths=0.5, zorder=3,
    )

    # All rows in df have valid n_TOV since we filtered in run_analysis
    if "n_TOV" in df.columns:
        axes[2].scatter(
            df["n_TOV"],
            np.power(10, df["logpc_TOV"]) / utils.MeV_fm_inv3_to_geometric,
            s=3, c='pink', alpha=0.9, zorder=4, label=r"$M_{\rm{TOV}}$"
        )
    
    if "n_14" in df.columns and "logpc_14" in df.columns:
        axes[2].scatter(
            df["n_14"],
            np.power(10, df["logpc_14"]) / utils.MeV_fm_inv3_to_geometric,
            s=3, c='yellow', alpha=0.9, zorder=5, label=r"$M_{1.4}$"
        )
        
    axes[2].set_xscale("log")
    axes[2].set_yscale("log")
    axes[2].set_xlim(1e-1, 3e1)
    axes[2].set_ylim(1e-2, 1e4)
    axes[2].set_xlabel("n")
    axes[2].set_ylabel("p")
    axes[2].legend(loc="upper left")

    fig.suptitle(title)
    plt.tight_layout()
    plt.show()

def _make_particle_fraction_plot(df: pd.DataFrame, title: str):
    fig, ax = plt.subplots(figsize=(8, 5))

    plot_curve_collection(
        df, "n_orig", "proton_fraction",
        ax=ax, colors="blue", alpha_below=0.05, alpha_above=0.05, linewidths=0.7, zorder=1,
    )
    plot_curve_collection(
        df, "n_orig", "DU_fraction",
        ax=ax, colors="red", alpha_below=0.08, alpha_above=0.08, linewidths=0.7, zorder=2,
    )
    
    if "Yp_n14" in df.columns and "Yp_nTOV" in df.columns:
        ax.scatter(df["n_14"], df["Yp_n14"], s=6, c="red", alpha=0.8, label=r"$Y_p$ at $M_{1.4}$", zorder=3)
        ax.scatter(df["n_TOV"], df["Yp_nTOV"], s=6, c="green", alpha=0.8, label=r"$Y_p$ at $M_{TOV}$", zorder=4)
        ax.legend(loc="upper right")
        
    ax.set_xlim(0, 9)
    ax.set_xlabel("n/nsat")
    ax.set_ylabel(r"$Y_p$ & $X_{DU}$")
    ax.set_title(title)
    plt.tight_layout()
    plt.show()

def _make_density_distribution_plot(df: pd.DataFrame):
    plt.figure(figsize=(6, 4))
    
    if "n_14" in df.columns:
        plt.hist(df["n_14"].dropna(), bins=50, alpha=0.5, label=r"$n_{1.4}$")
        
    if "n_TOV" in df.columns:
        tov_vals = df["n_TOV"].dropna()
        if len(tov_vals) > 0:
            plt.hist(tov_vals, bins=50, alpha=0.5, label=r"$n_{\rm{TOV}}$ (valid only)")
            
    # Add nbreak distribution
    if "n_trans_val" in df.columns:
        nbreak_vals = df["n_trans_val"].dropna()
        if len(nbreak_vals) > 0:
            plt.hist(nbreak_vals, bins=50, alpha=0.5, label=r"$n_{\rm{break}}$")

    plt.title("Central Density Distributions")
    plt.xlabel("n/nsat")
    plt.legend()
    plt.tight_layout()
    plt.show()

def _make_cs2_plot(df: pd.DataFrame, title: str):
    if "cs2_array" not in df.columns:
        return

    fig, ax = plt.subplots(figsize=(8, 5))

    plot_curve_collection(
        df, "n", "cs2_array",
        ax=ax, colors="gray", alpha_below=0.15, alpha_above=0.15, linewidths=1.0, zorder=1,
    )

    if "cs2_n14" in df.columns:
        ax.scatter(df["n_14"], df["cs2_n14"], s=10, c="yellow", marker="o", alpha=0.8, label=r"$c_s^2$ at $M_{1.4}$", zorder=3)
    if "cs2_nTOV" in df.columns:
        ax.scatter(df["n_TOV"], df["cs2_nTOV"], s=10, c="green", marker="s", alpha=0.8, label=r"$c_s^2$ at $M_{\rm{TOV}}$", zorder=4)
    if "n_durca_nsat" in df.columns and "cs2_ndurca" in df.columns:
        ax.scatter(df["n_durca_nsat"], df["cs2_ndurca"], s=15, c="red", marker="^", alpha=0.9, label=r"$c_s^2$ at $n_{\rm{dUrca}}$", zorder=5)

    ax.axhline(1.0, color='k', linestyle=':', label='Causality Limit')

    ax.set_xlim(0, 10)
    ax.set_ylim(0, 1.2)
    ax.set_xlabel(r"$n/n_{\rm{sat}}$")
    ax.set_ylabel(r"$c_s^2$")
    ax.set_title(title)
    ax.legend(loc="upper right")
    plt.tight_layout()
    plt.show()

def print_h5_structure(filepath: str | Path):
    display(Markdown(f"### HDF5 Structure Keys for `{filepath}`"))
    if not os.path.exists(filepath):
        print(f"File not found: {filepath}")
        return
        
    with h5py.File(filepath, "r") as f:
        if "posterior" not in f:
            print("No 'posterior' group found.")
            return
        post = f["posterior"]
        if "parameters" in post:
            print("--- Parameters ---")
            for k in post["parameters"].keys():
                print(f"  {k}: {post['parameters'][k].shape}")
        if "derived_eos" in post:
            print("--- Derived EOS ---")
            for k in post["derived_eos"].keys():
                print(f"  {k}: {post['derived_eos'][k].shape}")

def run_analysis(file_specs: list[dict]):
    r"""
    Run full set of plots for user-specified HDF5 runs without autosaving.
    """
    display(Markdown(f"## 🚀 Commencing analysis for {len(file_specs)} provided models..."))

    for run in file_specs:
        h5_path = run["path"]
        run_label = run["label"]
        
        display(Markdown(f"### 📄 Individual Run: `{run_label}`"))
        
        # 1. Load Data
        dataset_dict = prepare_eos_datasets([run], display_status=False)
        if run_label not in dataset_dict:
            display(Markdown(f"❌ Failed to load {h5_path}. Skipping..."))
            continue
            
        df = dataset_dict[run_label]
        
        # --- Filter to samples with valid n_TOV for ALL subsequent analysis ---
        total_raw = len(df)
        if "n_TOV_valid" in df.columns:
            df = df[df["n_TOV_valid"]].copy()
        
        # 2. Macroscopic Relations
        display(Markdown("**Macroscopic Relations:**"))
        _make_macroscopic_plot(df, f"{run_label}: Macroscopic")

        # 3. Particle Fractions
        display(Markdown("**Particle Fractions:**"))
        _make_particle_fraction_plot(df, f"{run_label}: Particle Fractions")
        
        # 4. Speed of Sound
        display(Markdown("**Speed of Sound:**"))
        _make_cs2_plot(df, f"{run_label}: Speed of Sound")

        # 5. Spaghetti Curves
        display(Markdown("**EOS & MR Spaghetti:**"))
        plot_spaghetti(df, f"{run_label}: EOS & MR Spaghetti")

        # 6. Parameter Histograms (from filtered DataFrame — no HDF5 raw read)
        display(Markdown("**Parameter Distributions:**"))
        _make_parameter_histograms(df, run_label)

        # 7. Density Distributions
        display(Markdown("**Central Density Distributions:**"))
        _make_density_distribution_plot(df)

        # 8. dUrca Summary Statistics
        display(Markdown("**dUrca Summary Statistics:**"))
        display_statistics_table(
            compute_durca_statistics(df),
            title=f"{run_label}: dUrca Statistics",
            total_raw=total_raw,
        )

# =============================================================================
# EXECUTING CUSTOM RUNS
# =============================================================================
if __name__ == "__main__":
    # USER EDIT ZONE
    USER_FILES = [
        {"path": "test_prior/Sk_prior/outdir/results.h5", "label": "Sk_prior"},
        {"path": "test_prior/MM_prior/outdir/results.h5", "label": "MM_prior"},
        {"path": "test_prior/Sk_prior_rand_nbreak/outdir/results.h5", "label": "Sk_prior_rand_nbreak"},
        {"path": "test_prior/MM_prior_rand_nbreak/outdir/results.h5", "label": "MM_prior_rand_nbreak"},
    ]
    
    if USER_FILES:
        run_analysis(USER_FILES)
        
        # Print HDF5 Structure for the last run in the list
        print_h5_structure(USER_FILES[-1]["path"])
    else:
        print("Please configure USER_FILES list with your h5 paths to run the analysis.")
```

```python
import h5py
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, Markdown

def plot_nbreak_histogram(file_specs):
    """
    Plots the distribution of the transition density (nbreak) 
    for a given list of HDF5 run specifications.
    """
    plt.figure(figsize=(10, 6), dpi=150)
    
    # Generate colors based on the number of files
    colors = plt.cm.tab10(np.linspace(0, 1, max(10, len(file_specs))))
    
    found_any = False
    
    for idx, run in enumerate(file_specs):
        path = run["path"]
        label = run["label"]
        
        try:
            with h5py.File(path, "r") as f:
                # Target the derived_eos group first, then fallback to parameters
                post = f.get("posterior/derived_eos")
                if post is None:
                    post = f.get("posterior/parameters")
                    
                if post is None:
                    continue
                    
                # Hunt for the exact key
                nbreak_data = None
                for key in ["nbreak", "n_break", "n_trans", "ntrans"]:
                    if key in post:
                        nbreak_data = post[key][:]
                        break
                        
                if nbreak_data is not None:
                    # Filter out NaNs and infinities
                    nbreak_clean = nbreak_data[np.isfinite(nbreak_data)]
                    
                    if len(nbreak_clean) == 0:
                        continue
                        
                    # Assuming the raw data is in fm^-3, convert to n/nsat
                    # (If it's already in nsat, just remove the / 0.16)
                    nbreak_nsat = nbreak_clean / 0.16
                    
                    # Plot filled histogram and outline
                    plt.hist(nbreak_nsat, bins=40, density=True, alpha=0.3, color=colors[idx], label=label)
                    plt.hist(nbreak_nsat, bins=40, density=True, histtype="step", lw=2, color=colors[idx])
                    
                    # Add median line
                    median_val = np.median(nbreak_nsat)
                    plt.axvline(median_val, color=colors[idx], linestyle="dashed", alpha=0.8)
                    
                    found_any = True
                else:
                    print(f"Warning: 'nbreak' key not found in {label}.")
                    
        except Exception as e:
            print(f"Error reading {label}: {e}")
            
    if found_any:
        plt.xlabel(r"$n_{\rm{break}}/n_{\rm{sat}}$", fontsize=14)
        plt.ylabel("Density", fontsize=14)
        plt.title("Transition Density Distribution ($n_{\\rm{break}}$)", fontsize=16)
        plt.grid(True, alpha=0.3)
        plt.legend(loc="upper right")
        plt.tight_layout()
        plt.show()
    else:
        plt.close()
        display(Markdown("❌ **No `nbreak` data found across any files! Check your HDF5 keys.**"))

# Execute the function using the USER_FILES list defined in the previous cell
plot_nbreak_histogram(USER_FILES)
```

```python
# =============================================================================
# DEBUG: Diagnose nbreak vs n_TOV discrepancy
# =============================================================================

def diagnose_nbreak_ntov_discrepancy(file_specs):
    """
    Compare the nbreak histogram (from HDF5) against the DataFrame's
    n_trans_val and n_TOV columns to diagnose why nbreak > n_TOV is rare
    despite many high nbreak values in the histogram.
    """
    import warnings
    warnings.filterwarnings("ignore", category=FutureWarning)

    for run in file_specs:
        label = run["label"]
        path = run["path"]

        # --- 1. Raw nbreak from HDF5 (same path as plot_nbreak_histogram) ---
        raw_nbreak = None
        src_group = "not found"
        with h5py.File(path, "r") as f:
            for grp_name in ["posterior/derived_eos", "posterior/parameters"]:
                grp = f.get(grp_name)
                if grp is None:
                    continue
                for key in ["nbreak", "n_break", "n_trans", "ntrans"]:
                    if key in grp:
                        raw_nbreak = grp[key][:]
                        src_group = f"{grp_name}/{key}"
                        break
                if raw_nbreak is not None:
                    break

        # --- 2. Load DataFrame to see what prepare_eos_dataframe produces ---
        df = prepare_eos_dataframe(path, display_status=False)

        # --- 3. Report ---
        display(Markdown(f"### 🔍 Discrepancy Diagnosis: `{label}`"))

        display(Markdown(
            f"**HDF5 nbreak source:** `{src_group}`  \n"
            f"**Raw samples in HDF5:** {len(raw_nbreak) if raw_nbreak is not None else 'N/A'}  \n"
            f"**Samples in DataFrame:** {len(df)}"
        ))

        # Summary of n_trans_val
        t = df["n_trans_val"]
        tov = df["n_TOV"]

        display(Markdown("**`n_trans_val` (nbreak in n/nsat):**"))
        print(f"  mean={t.mean():.1f}, median={t.median():.1f}, min={t.min():.1f}, max={t.max():.1f}")
        print(f"  NaN count: {t.isna().sum()} / {len(t)}")

        display(Markdown("**`n_TOV` (central density at TOV in n/nsat):**"))
        print(f"  mean={tov.mean():.1f}, median={tov.median():.1f}, min={tov.min():.1f}, max={tov.max():.1f}")
        print(f"  NaN count: {tov.isna().sum()} / {len(tov)}")
        print(f"  Rows where logpc_TOV <= -11.5 (filtered): {(df['logpc_TOV'] <= -11.5).sum()}")
        print(f"  Rows where logpc_TOV is NaN: {df['logpc_TOV'].isna().sum()}")

        # Comparison
        mask = t > tov
        # Also check with NaN-safe: exclude rows where either is NaN
        both_finite = t.notna() & tov.notna()
        mask_finite = (t[both_finite] > tov[both_finite])
        print(f"\n--- Comparison ---")
        print(f"  n_trans_val > n_TOV (all): {mask.sum()} / {len(df)}")
        print(f"  n_trans_val > n_TOV (both finite): {mask_finite.sum()} / {both_finite.sum()}")
        
        # If n_TOV is large, compare to nbreak prior
        print(f"\n  n_TOV >= 25: {(tov >= 25).sum()} / {len(tov)}")
        print(f"  n_TOV >= 20: {(tov >= 20).sum()} / {len(tov)}")
        print(f"  n_TOV >= 15: {(tov >= 15).sum()} / {len(tov)}")
        print(f"  n_TOV >= 10: {(tov >= 10).sum()} / {len(tov)}")

        # Check column names available
        trans_col_candidates = ["nbreak", "n_break", "n_trans", "ntrans", "n_match"]
        found_cols = [c for c in trans_col_candidates if c in df.columns]
        display(Markdown(f"**Transition column found:** `{found_cols}`  →  `{'found' if found_cols else 'NOT FOUND — fallback n_trans_val = n_TOV'}`"))

        plot_nbreak_histogram([run])

# Run the diagnostic
diagnose_nbreak_ntov_discrepancy(USER_FILES)
```
